Data Consistency & Derivation:

Deriving new metrics such as age at retirement and pension status.

```python
df['retirement_date']= pd.to_datetime(df['retirement_date'])
```

```python
df['dob']=pd.to_datetime(df['dob'])
```

```python
df[['retirement_date','dob']].info()
```

```python
df['age']=((df['retirement_date']-df['dob']).dt.days/365.25).astype(int)
```

```python
df
```

```python
df['pension_status']= df.apply(
    lambda df : 'Active' if df['active_flag'] == 1
    else 'Retired' if df['retirement_date'] < pd.Timestamp.today()
    else 'Inactive',
    axis= 1)
```

```python
df
```

```python
df[['retirement_date','pension_type','pension_status']]
```

Crucial Update: Re-calculating the years_of_service field based on the logic (retirement_date - join_date) to resolve discrepancies between recorded 
service years and actual timeline data.

```python
df['join_date']= pd.to_datetime(df['join_date'])
```

```python
df['years_of_service']=((df['retirement_date']-df['join_date']).dt.days/365.25).astype(int)
```

```python
df
```

Privacy Protection: Masking sensitive fields (e.g., beneficiary names, account
details).

```python
df['name']= df['name'].apply(
    lambda x : '*' * (len(x)-3)+x[-3:] if pd.notna(x) else x )
```

```python
df.to_csv('pension_dataset.csv', index=False)
```
